```python
import yt_dlp
```

```python
# The URL of the video you want to download
urls = ["https://www.youtube.com/watch?v=wkv9RhMmIzw&list=PL_VIYA-L9VnKSTAuwyPsyjVh4vQ_RekNe&index=69"]

# Specify the output directory
output_directory = "/Users/halilergul/Desktop/personal_projects/voice-data/voice_datasets"

# Configure yt-dlp options
ydl_opts = {
    # Specify the output path and template for filenames
    # This will save files to the Desktop with their original title and extension
    'outtmpl': f'{output_directory}/%(title)s.%(ext)s',

    # Options for extracting audio
    'format': 'bestaudio/best', # Choose the best audio quality
    'postprocessors': [{
        'key': 'FFmpegExtractAudio', # Use FFmpeg to extract audio
        'preferredcodec': 'wav',     # Convert audio to WAV format
        'preferredquality': '192',   # Optional: specify audio quality
    }],

    # Options for subtitles
    'writesubtitles': True,        # Download subtitles
    'subtitleslangs': ['tr'],      # Download Turkish subtitles
    'skip_download': False,        # Ensure the audio is downloaded (not just info extraction)
}

# Create a YoutubeDL object with the specified options and download the video(s)
try:
    with yt_dlp.YoutubeDL(ydl_opts) as ydl:
        error_code = ydl.download(urls)
        if error_code == 0:
            print(f"Successfully downloaded audio and Turkish subtitles to: {output_directory}")
        else:
            print(f"An error occurred. Error code: {error_code}")
except Exception as e:
    print(f"An exception occurred: {e}")
```
